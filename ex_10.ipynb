```python
valor = float(input("Digite o valor da compra: "))
pagamento = input("Forma de pagamento (avista/cartao): ")

if valor > 1000:
    desconto = 15
elif valor >= 500:
    desconto = 10
else:
    desconto = 5

if pagamento == "avista":
    desconto += 10

valor_final = valor - (valor * desconto / 100)

print(f"Desconto total: {desconto}%")
print(f"Valor final: R${valor_final:.2f}")
```
