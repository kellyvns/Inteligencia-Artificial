```python
celsius = float(input("Digite a temperatura em graus celsius: "))
fahrenheit = celsius * 9/5 + 32
kelvin = celsius + 273.15

if celsius < 15:
  clima = "frio"
elif celsius <= 28:
  clima = "ameno"
else:
  clima = "quente"

print(f"{celsius}°C = {fahrenheit}°F = {kelvin}K — Clima: {clima}")
```
