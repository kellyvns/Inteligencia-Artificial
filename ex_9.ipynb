```python
nota = float(input("Digite a nota final: "))
frequencia = float(input("Digite a frequência (%): "))

if frequencia < 75:
  print("Aluno reprovado por baixa frequência.")
else:
  if nota >= 6:
    print("Aluno aprovado!")
  elif nota >= 4:
    print("Aluno de recuperação.")
  else:
    print("Aluno reprovado.")
```
