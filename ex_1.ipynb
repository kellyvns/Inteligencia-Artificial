```python
nome = input("Nome: ")
idade = int(input("Idade: "))
cidade = input("Cidade: ")
eh_estudante = input("Você é estudante? (sim/nao): ").strip().lower() == "sim"

tem_meia_entrada = idade >= 18 and eh_estudante

print(f"{nome}, de {cidade}, {'tem' if tem_meia_entrada else 'não tem'} direito a meia-entrada.")
```
