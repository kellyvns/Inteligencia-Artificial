```python
peso = float(input("Digite o peso: "))
altura = float(input("Digite a altura: "))
idade = int(input("Digite a idade: "))

if idade < 18:
    print("Classificação de IMC não indicada para menores de idade")
else:
    imc = peso / (altura ** 2)

    if imc < 18.5:
        print("Abaixo do peso")
    elif imc < 25:
        print("Peso normal")
    elif imc < 30:
        print("Sobrepeso")
    else:
        print("Obesidade")
```
