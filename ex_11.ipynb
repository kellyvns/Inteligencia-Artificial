```python
a = float(input("Digite o primeiro número: "))
b = float(input("Digite o segundo número: "))
op = input("Digite a operação (+, -, *, /): ")

if op == "+":
    resultado = a + b
    print(f"Resultado: {resultado:.2f}")

elif op == "-":
    resultado = a - b
    print(f"Resultado: {resultado:.2f}")

elif op == "*":
    resultado = a * b
    print(f"Resultado: {resultado:.2f}")

elif op == "/":
    if b == 0:
        print("Erro: divisão por zero")
    else:
        resultado = a / b
        print(f"Resultado: {resultado:.2f}")

else:
    print("Operação inválida")
```
